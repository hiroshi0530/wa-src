リポジトリのルートで `uv sync --locked` を実行し、この Notebook のカーネルに `.venv` の Python、または登録済みの **wayama.io (Python 3.14)** を選んでください。

Run `uv sync --locked` from the repository root, then select the Python interpreter in `.venv` or the registered **wayama.io (Python 3.14)** kernel.

この Notebook は実行中の Python、共通ライブラリのインポート・バージョン、簡単な数値計算とグラフ表示を確認します。外部データは不要です。

This Notebook checks the active Python interpreter, imports and versions of the shared libraries, and a small calculation and plot. No external data is required.

## 1. Python と仮想環境 / Python and virtual environment

```python
import platform
import sys

print("Python:", platform.python_version())
print("Executable:", sys.executable)
print("Environment:", sys.prefix)
print("Virtual environment:", sys.prefix != sys.base_prefix)
print("OS:", platform.platform())
```

## 2. 共通ライブラリ / Shared libraries

各ライブラリをインポートしてからバージョンを表示します。エラーになる場合は、カーネルがルートの `.venv` を使っていることを確認し、`uv sync --locked` を再実行してください。

Each library is imported before its version is displayed. If an import fails, check that the kernel uses the root `.venv` and run `uv sync --locked` again.

```python
from importlib import import_module
from watermark import watermark

packages = {
    "ipykernel": "ipykernel",
    "ipython": "IPython",
    "ipywidgets": "ipywidgets",
    "jupyterlab": "jupyterlab",
    "matplotlib": "matplotlib",
    "networkx": "networkx",
    "numpy": "numpy",
    "pandas": "pandas",
    "polars": "polars",
    "scikit-learn": "sklearn",
    "scipy": "scipy",
    "seaborn": "seaborn",
    "sympy": "sympy",
    "watermark": "watermark",
}

for module_name in packages.values():
    import_module(module_name)

print(watermark(python=True, packages=",".join(packages)))
```

## 3. 計算とグラフ表示 / Calculation and plotting

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

x = sp.symbols("x")
print("Integral of sin(x) from 0 to pi:", sp.integrate(sp.sin(x), (x, 0, sp.pi)))

values = np.linspace(0, 2 * np.pi, 100)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(values, np.sin(values))
ax.set(xlabel="x", ylabel="sin(x)", title="Shared environment check")
ax.grid(alpha=0.3)
plt.show()
```
